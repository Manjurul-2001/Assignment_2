```python
#A2Q3(a)
import numpy as np 
import matplotlib.pyplot as plt 
from sympy import symbols , exp, sin, cos, simplify, lambdify, Matrix, diff, sqrt 

#Define symbols 
t = symbols('t') 

#Define two vector functions 
r1 = Matrix([exp(t), exp(t)*cos(t), exp(t)*sin(t)]) 
r2 = Matrix([2*cos(t), 3*sin(t), 0]) 

#Compute TNB Kappa and Tau 
def compute_TNB_kappa_tau(r) :
    r1_prime = diff(r, t) 
    r2_prime = diff(r1_prime, t)  
    r3_prime = diff(r2_prime, t)  

    T = simplify(r1_prime/sqrt(r1_prime.dot(r1_prime)))   

    T_prime = diff(T, t) 
    N = simplify(T_prime/sqrt(T_prime.dot(T_prime))) 

    B = simplify(T.cross(N)) 

    kappa = simplify(sqrt((r1_prime.cross(r2_prime)).dot((r1_prime.cross(r2_prime)))) / sqrt(r1_prime.dot(r1_prime))**3) 

    cross = r1_prime.cross(r2_prime)
    tau = simplify(cross.dot(r3_prime) / sqrt(cross.dot(cross))**2)  

    return T, N, B, kappa, tau 

#print T, N, B, kappa and tau for curve 1
T1, N1, B1, kappa1, tau1 = compute_TNB_kappa_tau(r1) 
print("\nCurve 1:") 
print("\nTangent T(t):", T1)
print("\nNormal N(t):", N1)
print("\nBinormal B(t):", B1)
print("\nCurvature κ(t):", kappa1) 
print("\nTorsion τ(t):", tau1) 

#print T, N, B, kappa and tau for curve 2
T2, N2, B2, kappa2, tau2 = compute_TNB_kappa_tau(r2) 
print("\nCurve 2:") 
print("\nTangent T(t):", T2)
print("\nNormal N(t):", N2)
print("\nBinormal B(t):", B2)
print("\nCurvature κ(t):", kappa2) 
print("\nTorsion τ(t):", tau2) 

#Print Curvature 
kappa1_func = lambdify(t, kappa1, 'numpy')
kappa2_func = lambdify(t, kappa2, 'numpy') 
t_val = np.linspace(0, 2*np.pi, 200) 
kappa1_val = kappa1_func(t_val)
kappa2_val = kappa2_func(t_val) 
plt.plot(kappa1_val, t_val, label="kappa1(t)") 
plt.plot(kappa2_val, t_val, label="kappa2(t)") 
plt.title("Printing Curvature kappa")
plt.legend() 
plt.grid() 
plt.show() 
```

```python
#A2Q4(c)
import numpy as np 
import matplotlib.pyplot as plt 
import sympy as sp 

x, y = sp.symbols("x, y") 
f1 = 4*x*y - x**4 - y**4
f2 = 4*x**2 * sp.exp(y) - 2*x**4 - sp.exp(4*y)

functions = [f1, f2] 
for idx, f in enumerate(functions, start = 1) :
    print(f"\nFunction {idx} : f(x,y) = {f}")

    #First partial derivative 
    fx = sp.diff(f,x) 
    fy = sp.diff(f,y) 

    #Solve for critical values 
    crit_points = sp.solve([fx, fy], (x, y), dict=True)
    print("Critical points are : ", crit_points) 

    #Compute partial second derivatives 
    fxx = sp.diff(fx, x) 
    fyy = sp.diff(fy, y) 
    fxy = sp.diff(fx, y) 

    #Classify each points
    for pt in crit_points :
        x0 = pt[x] 
        y0 = pt[y] 
        D = fxx.subs(pt)*fyy.subs(pt) - fxy.subs(pt)**2 
        fx_val = fxx.subs(pt)
        print(f"\nAt point ({x0},{y0}) : ") 
        print("\nDiscriminat = ", D) 

        if D > 0 : 
            real_fx_val = fx_val.evalf()
            if real_fx_val.is_real:
                if real_fx_val > 0:
                    print("→ Local Minimum")
                elif real_fx_val < 0:
                    print("→ Local Maximum")
                else:
                    print("→ Possibly saddle point or flat")
            else:
                print("→ f_xx is not real: cannot classify")
        elif D < 0 :
            print("Saddle Points")
        else :
            print("Inconclusive") 

#Graphical Verification 
def plot_surface(f_expr, title, xlim=(-3, 3), ylim=(-3, 3)):
    f_lambda = sp.lambdify((x, y), f_expr, 'numpy')
    X_vals = np.linspace(*xlim, 100)
    Y_vals = np.linspace(*ylim, 100)
    X, Y = np.meshgrid(X_vals, Y_vals)
    Z = f_lambda(X, Y)

    fig = plt.figure()
    ax = fig.add_subplot(111, projection='3d')
    ax.plot_surface(X, Y, Z, cmap='coolwarm')
    ax.set_title(title)
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_zlabel('f(x, y)')
    plt.tight_layout()
    plt.show() 

# Plot both functions
plot_surface(f1, "Function (i): $f(x,y) = 4xy - x^4 - y^4$")
plot_surface(f2, "Function (ii): $f(x,y) = 4x^2 e^y - 2x^4 - e^{4y}$", xlim=(-2, 2), ylim=(-1, 1))
```

```python
#A2Q5(a)
import numpy as np 
import matplotlib.pyplot as plt 
from sympy import symbols, lambdify, sqrt, Eq, pi, solve, acos, diff

#Define symbolic variables 
x, y, z, t = symbols("x y z t") 

#Define the ellipsoid function 
f = x**2 + 4*y**2 + z**2 

#Point on the ellipsiod 
x0, y0, z0 = 1, 2, 1
point = (x0, y0, z0) 

# === Part (i): tangent plane equation ===

#Compute gradient 
grad_f = [diff(f, var) for var in (x, y, z)] 

#Evaluate gradient at the point (1,2,1)
grad_f_at_a_point = [g.subs({x: x0, y: y0, z: z0}) for g in grad_f] 
a, b, c = grad_f_at_a_point

# Tangent plane equation: a(x - x0) + b(y - y0) + c(z - z0) = 0
# Expand to standard form: ax + by + cz = d
d = a*x0 + b*y0 + c*z0 
tangent_plane_equation = Eq(a*x + b*y + c*z, d) 

print("\nPart (i): Tangent Plane Equation:")
print(f"{a}*x + {b}*y + {c}*z = {d}") 

# === Part(ii) : Normal Line ===

#Parametric equation of the normal line 
x_line = x0 + a*t 
y_line = y0 + b*t 
z_line = z0 + c*t 

print("\nPart (ii): Parametric Equations of the Normal Line:")
print(f"x = {x_line}")
print(f"y = {y_line}")
print(f"z = {z_line}") 

# === Part (iii): Angle with xy-plane ===

# Normal vector to the plane: (a, b, c)
# Unit normal to xy-plane is (0, 0, 1)
# cos(θ) = |c| / ||(a, b, c)||

magnitude = sqrt ( a**2 + b**2 + c**2) 
cos_theta = abs(c) / magnitude 
angle_rad = acos(cos_theta)
angle_deg = angle_rad * 180 / pi

print(f"\nPart(iii) : Acute angle with xy-plane is {angle_deg.evalf():.2f} degrees.") 

# === Part (iv): Visualization ===

# 3D plot
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Plot the ellipsoid
u = np.linspace(0, 2*np.pi, 100)
v = np.linspace(0, np.pi, 100)
xe = np.sqrt(18) * np.outer(np.cos(u), np.sin(v))
ye = np.sqrt(18/4) * np.outer(np.sin(u), np.sin(v))
ze = np.sqrt(18) * np.outer(np.ones_like(u), np.cos(v)) 
ax.plot_surface(xe, ye, ze, color='lightblue', alpha=0.5, linewidth=0)

# Tangent plane surface near the point
xx, yy = np.meshgrid(np.linspace(-2, 4, 20), np.linspace(0, 4, 20))
zz = (-a*(xx - x0) - b*(yy - y0))/c + z0
ax.plot_surface(xx, yy, zz.astype(np.float64), color='orange', alpha=0.6)

# Plot the normal line
t_vals = np.linspace(-1, 1, 100)
line_x = x0 + float(a)*t_vals
line_y = y0 + float(b)*t_vals
line_z = z0 + float(c)*t_vals
ax.plot(line_x, line_y, line_z, color='red', label='Normal Line')

# Plot the point
ax.scatter([x0], [y0], [z0], color='black', s=50, label='Point (1, 2, 1)')

# Labels
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Ellipsoid with Tangent Plane and Normal Line')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
#A2Q5(b)
from sympy import symbols, Eq, solve 

#Define Symbols 
x, y, z, lmbd = symbols("x y z lmbd") 

#Define gradient equations from ∇T = λ∇g 
eq1 = Eq(16*x, lmbd * 8*x)
eq2 = Eq(4*z, lmbd * 2*y)
eq3 = Eq(4*y - 16, lmbd * 8*z) 

# Constraint (ellipsoid surface)
constraint = Eq(4*x**2 + y**2 + 4*z**2, 16) 

#Solve the system 
solutions = solve([eq1, eq2, eq3, constraint], (x, y, z, lmbd), dict = True)  

# Evaluate T at the solutions to find max
T = lambda sol: 8*sol[x]**2 + 4*sol[y]*sol[z] - 16*sol[z] + 600 
max_temp = -float('inf')
max_point = None 

for sol in solutions :
    try:
        temp = T(sol) 
        if temp > max_temp :
            max_temp = temp 
            max_point = sol 
    except: 
        continue

print("\n hottest point in the ellipsiod surface is: ", max_point) 
print("\nMaximum temprature: ", max_temp, "\n")
```

```python
#A2Q7(c)
import sympy as sp 

# Define symbolic variables
r, h, rp, theta, z = sp.symbols('r h r_prime theta z') 

# Define the integrand in polar coordinates: r'^3
integrand = rp**3

inner_integration = sp.integrate(integrand, (rp, 0, r))
middle_integration = sp.integrate(inner_integration, (theta, 0, 2*sp.pi)) 
mass = sp.integrate(middle_integration, (z, -h/2, h/2)) 
mass_simplified = sp.simplify(mass) 

print(f"\nThe mass of the cylinder is : {mass_simplified} \n")
```

```python
#A2Q7(d)
import sympy as sp 

#Define symbols 
x, y = sp.symbols("x y") 

#Define the vector field 
F = sp.Matrix([sp.exp(y), x*sp.exp(y)]) 
P, Q = F[0], F[1] 

dQ_dx = sp.diff(Q, x) 
dP_dy = sp.diff(P, y) 

if dQ_dx == dP_dy :
    print("\nThe field is conservative.") 
else :
    print("\nThe field is not conservative.") 

#Compute the potential function 
phi = sp.integrate(P, x) 
dP_dy = sp.diff(phi, y) 
print("\nThe potential function is phi(x, y) : ", phi) 

#Compute the work done 
phi_func = sp.lambdify((x,y), phi) 
work = phi_func(-1, 0) - phi_func(1, 0) 
print("The work done is given by : ", work)
```

```python
#A2Q8(a)
from scipy.integrate import dblquad
import numpy as np 

# Define the integrand 
def integrand(r, theta):
    return r**3 

# Correct limits: inner limits must be functions of theta
work_done, error = dblquad(integrand, 0, 2*np.pi, lambda theta: 0, lambda theta: 1)

# Multiply by 3 due to Green's Theorem
work_done *= 3

print("\nThe work done is given by", work_done) 
```

```python
#A2Q8(b)
import numpy as np
from scipy.integrate import dblquad

#Define the integrand 
def integrand (phi, theta ): 
    return ((np.cos(theta)*np.sin(phi))**2) * (np.sin(phi)) 

result, error = dblquad(integrand, 0, 2*np.pi, 0, np.pi) 

print("Surface integral of x^2 over the unit sphere is:", result)
```

```python
#A2Q8(c)
import numpy as np 
from scipy.integrate import tplquad

#Define the integrand (divergence*r)
def integrand (r, theta, z) :
    return (3*r**2 + 2*z)*r 

# f(r, theta, z) : use limit z, theta, r 
result, error = tplquad(integrand, 0, 2, 0, 2*np.pi, 0, 3) 
print("\nThe outward flux is : ", result)
```

```python
#A2Q8(d)
import numpy as np 
from scipy.integrate import quad 

#Line integral part 
def integrand(theta) :
    return 12*np.cos(theta)**2 
line_integral, _ = quad(integrand, 0, 2* np.pi) 

#surface ingtegral part(since curl is constant and area vector is 4*pi in k direction) 
surface_integral = 3*4*np.pi #dot  product with 3k^

if line_integral == surface_integral :
    print("\nStoke's Theorem verified.") 
else:
    print("\nDoes not satisfy stokes theorem") 

print(f"\nThe line integral is : {line_integral}")
print(f"\nThe surface integral is : {surface_integral}") 
```
